есть отель, для каждого посетителя известно время заезда и отъезда. Найти день, имеющий момент с макс числом посетителей. Судя по предложенным данным даты квантуются по дням. Формат даты: год-месяц-день

```python
def translate(cortege_data):
    data1 = cortege_data[0]
    data1 = int(data1[0:4]) * 12 * 31 + int(data1[5:7]) * 31 + int(data1[8:10])
    data2 = cortege_data[1]
    data2 = int(data2[0:4]) * 12 * 31 + int(data2[5:7]) * 31 + int(data2[8:10])
    return data1, data2

def data_reads(cortege_data):
    reads = []
    for i in cortege_data:
        data1, data2 = translate(i)
        reads.append((data1, data2))

    start = min(data1 for data1, data2 in reads)
    end = max(data2 for data1, data2 in reads)

    coverage = [0] * (end - start + 1)

    for data1, data2 in reads:
        for pos in range(data1, data2 + 1):
            coverage[pos - start] += 1   

    return coverage, start
```

```python
def number_to_date(n):
    # обратный перевод числа в строку 'ГГГГ-ММ-ДД'
    # месяц и день начинаются с 1, а не с 0, поэтому сначала вычитаем 31 + 1
    year, rest = divmod(n - 31 - 1, 12 * 31)   # divmod(a, b) возвращает (a // b, a % b)
    month, day = divmod(rest, 31)
    return f'{year:04}-{month + 1:02}-{day + 1:02}'

tests = [
    [("2024-09-15", "2024-09-15"), ("2024-09-14", "2024-09-21")],
    [("2024-09-15", "2024-09-15")],
    [("2024-09-01", "2024-09-05"), ("2024-09-03", "2024-09-10"), ("2024-09-04", "2024-09-04"),
     ("2024-09-06", "2024-09-08"), ("2024-09-07", "2024-09-12")],
    [("2024-09-01", "2024-09-05"), ("2024-09-05", "2024-09-07"), ("2024-09-05", "2024-09-05")],
    [("2024-09-01", "2024-09-02"), ("2024-09-05", "2024-09-05")],
    [("2024-09-28", "2024-10-03"), ("2024-09-30", "2024-10-01")],
    [("2023-12-30", "2024-01-02"), ("2024-01-01", "2024-01-01")],
]

for guests in tests:
    coverage, start = data_reads(guests)
    best = coverage.index(max(coverage))      # index возвращает ПЕРВУЮ позицию с таким значением
    print(number_to_date(start + best), 'посетителей:', max(coverage))
```
