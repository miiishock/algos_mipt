теперь чтобы перевернуть список достаточно лишь поменять направленность стрелок и голову переназначить

```python
class LinkedList:
    def __init__(self):
        self.head = None        

    def print_list(self):
        work_dupl = self.head
        while work_dupl != None:
            print(work_dupl.data)
            work_dupl = work_dupl.next  
    
    def append_node(self, data):
        if self.head == None:
            self.head = Node(data) 
        else:
            work_dupl = self.head
            while work_dupl != None:
                end = work_dupl
                work_dupl = work_dupl.next
            end.next = Node(data) 

    def reverse(self):
        prev = None              
        curr = self.head         
        while curr:
            nxt = curr.next      
            curr.next = prev     
            prev = curr          
            curr = nxt           
        self.head = prev         
```

```python
lst3 = LinkedList()
for i in range(1, 6):
    lst3.append_node(i)
print('до разворота:')
lst3.print_list()
lst3.reverse()
print('после разворота:')
lst3.print_list()
```
